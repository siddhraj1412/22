# FIFA-19 EDA

## 1. importing library

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Loading the dataset

```python
df=pd.read_csv('/kaggle/input/fifa19eda/fifa_eda.csv')
df.head()
```

## 3.Basic

```python
df.shape # to check shape of the dataset
```

which means it has 18207 player data and 18 feature

```python
df.info() # to get the basic info about the dataset
```

```python
df.describe() # to get detail info about the dataset
```

```python
df.isnull().sum() #to count total null values
```

There are some null values in the dataset 
- club : which means player doesnt play for any club
- value: which means player is free agent
- International Reputation : which means the player doesnt have played for his country so the reputation is null
- Skill Moves : which means player doesnt have any skill moves
- Contract Valid Until : which means player doesnt have any contract to be valid so its value is null


```python
df.dropna(inplace=True) # to delete all the rows that have null values
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df.duplicated().sum() # to check the duplicate values
```

which means it doesn't have any duplicate values

```python
df.nunique() # to get all the unique values for each columns
```

## 4.Age

```python
df['Age'].hist(edgecolor='black')
```

which shows us the overall distribution about the age feautre of our data

## 5.Nationality

```python
national = df['Nationality'].value_counts()[:10]
national.plot.bar(color='darkblue')
plt.show()
```

which shows us most of the players are from england and continant wise from europe 

## 6.Overall 

```python
df['Overall'].hist(figsize=(10,8),edgecolor='black')
plt.show()
```

which shows us most of the players rating are between 60 to 70 and they are small amount of people who has rating of 90+

## 7.Potential 

```python
df['Potential'].hist(figsize=(10,8),edgecolor='black',color='yellow',bins=8)
plt.show()
```

which shows us most of the players potential is between 65 to 75 and there are small amount of player who has potential of 90+ 

## 8.Club 

```python
national = df['Club'].value_counts()[:40]
national.plot.bar(color='red',figsize=(12,8))
plt.show()
```

which shows us there are nearly 25 clubs they have the most and nearly same amount of players in their club

## 9.Skill moves

```python
sns.countplot(x='Skill Moves',data=df)
plt.show()
```

which shows us the number of players who has that specific skill moves

## 10.prefered foot

```python
sns.countplot(x='Preferred Foot',data=df)
plt.show()
```

which shows us there are more right footed players then left

```python
plt.pie(df['Preferred Foot'].value_counts(),labels=['right','left'],autopct="%1.1f%%")
plt.show()
```

## 11.Internation Reputation

```python
sns.countplot(x='International Reputation',data=df)
plt.show()
```

which gives us the idea about the interanation reputation distribution amongst player and it also shows that most player has internation reputation of one

## 12.Joined

```python
df['Joined'].hist()
plt.xlabel('year')
plt.ylabel('player count')
plt.title('player joining year distribution')
plt.show()
```

which shows us player joining year distribution 

## 13.Height and weight

```python
df['Height'].hist()
plt.xlabel('year')
plt.ylabel('player count')
plt.title('player joining year distribution')
plt.show()
```

```python
df['Weight'].hist()
plt.xlabel('year')
plt.ylabel('player count')
plt.title('player joining year distribution')
plt.show()
```

where each graph shows us the weight and height distribution from the data

## 14.Value relations with other features

```python
df.plot.scatter(x='Value',y='Wage')
plt.xlabel('value')
plt.ylabel('wage')
plt.title('Value-wage relation')
plt.show()
```

which shows the relation between the value and wage

```python
df.plot.scatter(x='Age',y='Value')
plt.xlabel('Age')
plt.ylabel('value')
plt.title('age-value relation')
plt.show()
```

which shows us value realtion with age

```python
plt.figure(figsize=(9,5))
df.plot.scatter(x='Position',y='Value',figsize=(10, 5))
plt.xlabel('Position')
plt.ylabel('value')
plt.title('Position-value relation')
plt.tight_layout()
plt.show()
```

which shows us position-value relation where we can see that lw and rf has most value compared to others and lwb has lowest value

```python
df.plot.scatter(x='Release Clause',y='Value')
plt.xlabel('Release Clause')
plt.ylabel('value')
plt.title('Release Clause-value relation')
plt.tight_layout()
plt.show()
```

which shows us the Release Clause-value relation from the data
